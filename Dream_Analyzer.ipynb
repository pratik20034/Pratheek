```python
import google.generativeai as genai

# ✅ Reset input if it's overridden
input = __builtins__.input

# 🔐 Configure Gemini API
genai.configure(api_key="AIzaSyAy-TuAv5FkXBDE9rq9jxJdpWOjfhp4gOE")

# 💭 Initialize the model
model = genai.GenerativeModel("gemini-2.5-flash")

print("🌙 Dream Analyzer Bot 💤")
print("Describe your dream below (type 'exit' to quit)\n")

while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
        print("Goodbye! Sleep well 🌌")
        break

    prompt = f"""
    You are a dream analysis expert.
    Analyze the following dream in psychological and symbolic terms.
    Also explain what emotions or subconscious thoughts might be reflected.

    Dream: {user_input}
    """

    response = model.generate_content(prompt)
    print("\n🔮 Analysis:\n", response.text, "\n")
```
