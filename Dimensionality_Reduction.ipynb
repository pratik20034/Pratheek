#### 2. Use dimensionality reduction (e.g., PCA or t-SNE) to visualize word embeddings for

 Q1. Select 10 words from a specific domain (e.g., sports, technology) and visualize their embeddings.Analyze clusters and relationships. Generate contextually rich outputs using embeddings. Write a program to generate 5 semantically similar words for a given input.




1.  **Word Embeddings**:
    
    *   The program loads **Google's Word2Vec (300-dimensional) embeddings**.
        
    *   It extracts vectors for **10 technology-related words**.
        
2.  **Dimensionality Reduction**:
    
    *   **PCA** is used to reduce dimensions from **300 to 10**.
        
        
3.  **Visualization**:
    
    *   The embeddings are **plotted on a 2D graph**.
        
    *   Words are displayed with scatter points and labels.
        
4.  **Semantic Similarity**:
    
    *   The function get\_similar\_words() retrieves **5 most similar words** for an input word using most\_similar() from Word2Vec.

```python
import gensim.downloader as api
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Load the Word2Vec pre-trained model
model = api.load("word2vec-google-news-300")

# Select 10 words from the technology domain
tech_words = ["AI", "algorithm", "computer", "data", "machine", "neural", "software", "hardware", "programming", "network"]

# Get word vectors for the selected words
word_vectors = np.array([model[word] for word in tech_words])

# Apply PCA for dimensionality reduction to 2D
pca = PCA(n_components=2)
word_vectors_pca = pca.fit_transform(word_vectors)

# Plot the embeddings
plt.figure(figsize=(10, 6))
for i, word in enumerate(tech_words):
    x, y = word_vectors_pca[i]
    plt.scatter(x, y)
    plt.text(x + 0.05, y + 0.05, word, fontsize=12)

plt.title("PCA Visualization of Word Embeddings (Technology Domain)")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.show()

# Function to generate 5 semantically similar words
def get_similar_words(word, topn=5):
    if word in model:
        similar_words = model.most_similar(word, topn=topn)
        return [w[0] for w in similar_words]
    else:
        return f"Word '{word}' not in vocabulary."

# Example: Get similar words for "AI"
input_word = "AI"
similar_words = get_similar_words(input_word)
print(f"Words similar to '{input_word}': {similar_words}")
```
